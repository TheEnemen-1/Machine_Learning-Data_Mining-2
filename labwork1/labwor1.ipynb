# Feature Breakdown

## Metadata & Track

- track_id: Spotify unique identifier for each track. 
- artists: Name(s) of the performing artist(s).
- album_name: Name of the album containing the track.
- track_name: Title of the track.
- popularity: Value from 0 to 100 indicating how popular the track is on Spotify.
- duration_ms: Duration of the song in milliseconds.
- explicit: Boolean (True/False) indicating whether the lyrics contain explicit content.
- track_genre: The music genre label (e.g., acoustic, pop, rock, metal, edm, jazz).

## Audio & Musical Characteristics

- danceability (0.0 - 1.0): How suitable a track is for dancing based on tempo, rhythm stability, beat strength.
- energy (0.0 - 1.0): Perceptual measure of intensity, speed, and noise (e.g., death metal has high energy).
- loudness (dB): Overall volume of the track in decibels (typically ranges from -60 to 0 dB).
- speechiness (0.0 - 1.0): Detects the presence of spoken words (podcasts/talk > 0.66; rap ~ 0.33 - 0.66).
- acousticness (0.0 - 1.0): Confidence score that the track is acoustic (non-electronic).
- instrumentalness (0.0 - 1.0): Likelihood that the track contains no vocals.
- liveness (0.0 - 1.0): Detects the presence of an audience (live performance if > 0.8).
- valence (0.0 - 1.0): Musical positiveness conveyed by the song (high = cheerful/happy, low = sad/angry).

## Musical Structure 

- key: Pitch class notation (0 = C, 1 = C♯/D♭, ..., 11 = B; -1 if not detected).
- mode: Modality of the track (1 = Major, 0 = Minor).
- tempo: Estimated speed in Beats Per Minute (BPM).
- time_signature: Meter of the track (e.g., 3 for 3/4, 4 for 4/4).

# Consequence for data and machine learning 

1. Missing Values

- Algorithm Failure: Most standard ML models (Linear Regression, SVM, Neural Networks, k-NN) in Scikit-Learn will throw an immediate error if fed NaN values.
- Biased Imputation: Dropping rows reduces sample size, while naive imputation (e.g., filling with mean or mode) can artificially deflate variance or distort feature correlations.

2. Duplicate Records

- Data Leakage (Critical): If duplicate tracks are present and you split data into Train and Test sets, the exact same song can appear in both sets. This leads to artificially inflated test accuracy (overfitting) that will fail in real-world deployment.
- Distorted Weighting: Songs assigned to 3–4 genres will have 3–4x higher weight in loss functions and statistical summaries than single-genre songs.

3. Incorrect & Anomalous Values

- Distorts Descriptive Statistics: False zeros (e.g., tempo = 0 or loudness = -60) pull down the sample mean and inflate the standard deviation.
- Misleads Distance-based Algorithms: Algorithms relying on Euclidean distance (like k-Means clustering or k-NN) will treat tempo = 0 as a genuine physical property, grouping silent/corrupted files together erroneously.

4. Inconsistencies

- High Cardinality in One-Hot Encoding: Inconsistent artist names (e.g., "Artist A" vs "Artist A; Artist B") create redundant, sparse categorical features, increasing model complexity without adding value.
- Inaccurate Aggregations: Grouping by artist to compute metrics (e.g., average popularity per artist) will separate collaborations from solo tracks, yielding fragmented insights.

5. Outliers

- Distorts Linear Models: Ordinary Least Squares (OLS) minimizes squared errors, meaning extreme outliers (like a 2-hour track) exert disproportionate leverage on the regression line.
- Distorts Feature Scaling: Standardization (StandardScaler) or Min-Max normalization will be compressed because the minimum or maximum values are extreme, squashing the remaining 99% of normal data into a tiny band near 0.

# Data cleanning

# Data cleaning strategy and methodology

| Issue | Detection in Dataset | Strategy & Handling Rationale |
| --- | --- | --- |
| Missing Values (NaNs) | Exactly 1 row has `NaN` in `artists`, `album_name`, and `track_name`. | Drop the row (`dropna`): Because missing metadata cannot be meaningfully imputed, and 1 row out of 114,000 represents only 0.0008% of the data, dropping it causes zero loss of statistical power. |
| Duplicate Records | 24,259 duplicate `track_id` rows exist because the same song is tagged under multiple genres or appeared on different albums. | Deduplicate on `track_id` (`drop_duplicates`): Keep the first occurrence. Keeping duplicate tracks causes data leakage between train/test splits and artificially overweights multi-genre songs. |
| Incorrect Values | 157 tracks have `tempo == 0.0`, and 2 tracks have `duration < 10s` (silent audio, broken clips). | Filter out invalid rows: Songs with 0 BPM represent audio detection errors or silent files. We filter out `tempo > 0` and `duration_ms >= 10,000`. |
| Outliers | Extreme values in `duration_ms` (ambient/podcast recordings > 30 mins) and `loudness` (< -40 dB). | IQR (Interquartile Range) or Domain Bounds: Trim extreme `duration_ms` (> 10–15 mins) and extreme silence/distortion to prevent skewing linear models and feature scaling. |
| Irrelevant Column | `Unnamed: 0` is simply an old row index. | Drop the column (`drop(columns=['Unnamed: 0'])`). |

## Code

```python
import pandas as pd
import numpy as np

# 1. Load dataset as DataFrame
df_raw = dataset['train'].to_pandas()
df_clean = df_raw.copy()

print(f"Original shape: {df_clean.shape}")

# -------------------------------------------------------------
# STEP 1: Drop irrelevant index column
# -------------------------------------------------------------
if 'Unnamed: 0' in df_clean.columns:
    df_clean = df_clean.drop(columns=['Unnamed: 0'])

# -------------------------------------------------------------
# STEP 2: Handle Missing Values
# -------------------------------------------------------------
# Drop the 1 row with missing text values
df_clean = df_clean.dropna(subset=['artists', 'album_name', 'track_name'])

# -------------------------------------------------------------
# STEP 3: Handle Duplicate Records
# -------------------------------------------------------------
# Remove duplicate tracks by track_id, keeping the first occurrence
df_clean = df_clean.drop_duplicates(subset=['track_id'], keep='first')

# -------------------------------------------------------------
# STEP 4: Handle Incorrect / Anomalous Values
# -------------------------------------------------------------
# Filter out tracks with tempo == 0 and duration < 10 seconds (corrupted / silent tracks)
df_clean = df_clean[(df_clean['tempo'] > 0) & (df_clean['duration_ms'] >= 10000)]

# -------------------------------------------------------------
# STEP 5: Handle Outliers (IQR Method on duration_ms & loudness)
# -------------------------------------------------------------
def remove_iqr_outliers(data, column, factor=1.5):
    q25 = data[column].quantile(0.25)
    q75 = data[column].quantile(0.75)
    iqr = q75 - q25
    lower_bound = q25 - factor * iqr
    upper_bound = q75 + factor * iqr
    return data[(data[column] >= lower_bound) & (data[column] <= upper_bound)]

# Filter extreme durations and loudness
df_clean = remove_iqr_outliers(df_clean, 'duration_ms', factor=2.5)
df_clean = remove_iqr_outliers(df_clean, 'loudness', factor=2.5)

# Reset index
df_clean = df_clean.reset_index(drop=True)

print(f"Cleaned shape: {df_clean.shape}")
```

## Challenge encoutered in data intergration

There are four major challenges arise:

###  Mismatched Identifiers

- Collaborations and Multi-Artists: In this dataset, the artists field often contains multiple artists separated by semicolons (e.g., "David Guetta; Bebe Rexha"). An external artist table typically lists artists individually ("David Guetta" and "Bebe Rexha"). Simple exact string matching fails unless custom parsing is applied.
- Special Characters & Accents: Names with diacritics, accents, or symbols (e.g., Beyoncé, AC/DC, P!nk, Björk) frequently cause join mismatches between different sources.

### Granularity Mismatches

- Track vs. Artist Level: One artist has hundreds of tracks, while a single track can have multiple contributing artists. Merging on artist can either cause data duplication or force aggregations (like taking the average or maximum follower count among collaborators).
- Track vs. Album Level: Albums have distinct release dates and versions (Deluxe, Remastered), which can dilute track-level uniqueness.

### Temporal Inconsistency & Freshness

- Dynamic vs. Static Attributes: Audio features (tempo, key) are permanently static. However, popularity and followers change daily. Merging a 2021 track list with 2024 artist popularity data creates a temporal misalignment (e.g., a viral breakout artist in 2024 will appear to have had millions of followers when their 2020 song was released).

### Missing Keys & Asymmetric Coverage (Sparsity)

- Many indie, obscure, or regional tracks in the 114,000-row dataset will not have matching records in standard external databases (such as Billboard Top 100 or Genius Lyrics), resulting in significant NaN / null generation on left joins.

# Data transformation techniques

## Transformatoin strategy 

| Feature Category | Features | Applied Technique | Why This Technique is Appropriate |
| :--- | :--- | :--- | :--- |
| **Wide-range Numerical** | `tempo`, `loudness` | **Standardization (`StandardScaler`)** | Centers data around mean = 0, standard deviation = 1 ($z = \frac{x - \mu}{\sigma}$). Prevents features measured in large units (e.g., tempo ~ 120 BPM) from dominating features on smaller scales in distance/gradient algorithms. |
| **Skewed Numerical** | `duration_ms` | **Log Transform + Standardization** | Duration is heavily right-skewed. Taking $\log_{10}(\text{duration})$ compresses extreme long tracks and approximates a normal (Gaussian) distribution, followed by standard scaling. |
| **Already Bounded Numerical** | `danceability`, `energy`, `valence`, `speechiness`, `acousticness`, `instrumentalness`, `liveness` | **Min-Max Normalization / Retain $[0, 1]$** | Spotify's audio analysis API already normalizes these metrics to the $[0.0, 1.0]$ interval. Keeping them as-is or applying light Min-Max scaling preserves their physical probabilistic meaning. |
| **Binary Categorical** | `explicit` | **Binary Mapping (`0 / 1`)** | Converts boolean `True`/`False` into numeric integers `1` and `0` without expanding dimensionality. (`mode` is already binary 0/1). |
| **Cyclical Categorical** | `key` (0 to 11) | **Cyclical Encoding ($\sin / \cos$)** | Musical keys are cyclical: key 11 (B) is adjacent to key 0 (C). Linear numbers make key 11 seem maximally distant from key 0. Sine/Cosine encoding preserves this circular geometry. |
| **Nominal Categorical (Target)** | `track_genre` (114 classes) | **Label Encoding** *(for target)* or **One-Hot Encoding** *(for inputs)* | If `track_genre` is the prediction target, `LabelEncoder` maps 114 classes to integers `0`–`113`. For input features, One-Hot Encoding avoids false ordinal rank. |


## Code for data transformation

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder

# Work on a copy of the cleaned dataset
df_trans = df_clean.copy()

# ==============================================================================
# 1. LOG TRANSFORMATION (Handling Right-Skewness)
# ==============================================================================
# duration_ms has high variance and positive skewness; log transform normalizes it
df_trans['duration_log'] = np.log1p(df_trans['duration_ms'])

# ==============================================================================
# 2. STANDARDIZATION (Z-score scaling to mean=0, std=1)
# ==============================================================================
std_scaler = StandardScaler()
cols_to_standardize = ['loudness', 'tempo', 'duration_log']

df_trans[[f"{col}_std" for col in cols_to_standardize]] = std_scaler.fit_transform(
    df_trans[cols_to_standardize]
)

# ==============================================================================
# 3. MIN-MAX NORMALIZATION (Bounded to [0, 1])
# ==============================================================================
# Ensure audio ratio features are strictly bounded within [0, 1]
minmax_scaler = MinMaxScaler()
ratio_features = [
    'danceability', 'energy', 'speechiness', 
    'acousticness', 'instrumentalness', 'liveness', 'valence'
]
df_trans[[f"{col}_norm" for col in ratio_features]] = minmax_scaler.fit_transform(
    df_trans[ratio_features]
)

# ==============================================================================
# 4. CATEGORICAL ENCODING
# ==============================================================================
# A. Binary Encoding for 'explicit'
df_trans['explicit_encoded'] = df_trans['explicit'].astype(int)

# B. Cyclical Encoding for Musical 'key' (0 to 11 pitch classes)
# Preserves the circular relationship (key 11 is neighbor to key 0)
df_trans['key_sin'] = np.sin(2 * np.pi * df_trans['key'] / 12.0)
df_trans['key_cos'] = np.cos(2 * np.pi * df_trans['key'] / 12.0)

# C. Label Encoding for 'track_genre' (classification target)
genre_encoder = LabelEncoder()
df_trans['genre_encoded'] = genre_encoder.fit_transform(df_trans['track_genre'])

# D. One-Hot Encoding for 'time_signature' (e.g., 3/4, 4/4, 5/4)
time_sig_dummies = pd.get_dummies(df_trans['time_signature'], prefix='time_sig', drop_first=True, dtype=int)
df_trans = pd.concat([df_trans, time_sig_dummies], axis=1)

print("Transformations completed successfully!")
```

## Summary of Benefits for Machine Learning

- Equal Feature Weighting in Distance Calculations: In algorithms like k-Nearest Neighbors (k-NN), k-Means, or SVMs, raw duration_ms (values in hundreds of thousands) would completely overwhelm danceability (values between 0 and 1). Standardization puts all dimensions on an equal scale.
- Accelerated Gradient Descent Convergence: Linear Regression, Logistic Regression, and Neural Networks converge significantly faster and avoid numerical instability when feature inputs are centered at zero with unit variance.
- Valid Geometric Representation of Music: Cyclical sine/cosine encoding allows the machine learning model to realize that Key 11 (𝐵) and Key 0 (𝐶) are musically one semitone apart rather than 11 steps apart.

# Statistical Analysis: Mean, Variance, Covariance & Correlation


## 1. Mathematical Formulas for Numerical Features

### a. Sample Mean (Giá trị trung bình mẫu)
Measures the central tendency of a numerical feature:
$$\bar{x} = \frac{1}{N} \sum_{i=1}^{N} x_i$$

### b. Sample Variance (Phương sai mẫu)
Measures the degree of spread or dispersion of data points around the mean:
$$s^2 = \frac{1}{N-1} \sum_{i=1}^{N} (x_i - \bar{x})^2$$

### c. Sample Covariance (Hiệp phương sai mẫu)
Measures the joint linear variability between two continuous variables $X$ and $Y$:
$$\text{Cov}(X, Y) = \frac{1}{N-1} \sum_{i=1}^{N} (x_i - \bar{x})(y_i - \bar{y})$$
- $\text{Cov}(X, Y) > 0$: $X$ and $Y$ tend to increase together.
- $\text{Cov}(X, Y) < 0$: As $X$ increases, $Y$ tends to decrease.
- Scale-dependent (depends on units of measurement of $X$ and $Y$).

### d. Pearson Correlation Coefficient (Hệ số tương quan Pearson)

$$r_{XY} = \frac{\text{Cov}(X, Y)}{s_X \cdot s_Y} = \frac{\sum_{i=1}^{N} (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{N} (x_i - \bar{x})^2 \sum_{i=1}^{N} (y_i - \bar{y})^2}}$$

## 2. Analyzing Categorical Features

### Univariate Statistical Measures (Single Categorical Variable)
- **Mode:** The single most common category (central tendency).
- **Frequency & Relative Proportion:** Absolute count ($N_k$) and percentage share ($p_k = N_k / N$).
- **Shannon Entropy ($H$):** Measures the diversity/uncertainty across categories:
$$H(X) = - \sum_{k=1}^{K} p_k \log_2(p_k)$$
where $p_k = \frac{N_k}{N}$ is the relative proportion of category $k$.

### Bivariate Measures (Association Between Two Categorical Variables)
- **Contingency Table (Cross-Tabulation):** Matrix of joint frequencies between two categories (`pd.crosstab`).
- **Chi-Square Test of Independence ($\chi^2$):** Tests whether two categorical features are statistically independent:
$$\chi^2 = \sum_{i=1}^{r} \sum_{j=1}^{c} \frac{(O_{ij} - E_{ij})^2}{E_{ij}}$$
where $O_{ij}$ is the observed frequency and $E_{ij} = \frac{R_i \cdot C_j}{N}$ is the expected frequency under independence.
- **Cramér's V ($V$):** Normalized effect size measuring association strength bounded in $[0, 1]$:
$$V = \sqrt{\frac{\chi^2}{N \cdot \min(r - 1, c - 1)}}$$

### Categorical vs. Numerical Association
- **Point-Biserial Correlation ($r_{pb}$):** Measures correlation between a binary categorical feature (e.g., `explicit` $0/1$) and a continuous feature (e.g., `popularity`):
$$r_{pb} = \frac{\bar{X}_1 - \bar{X}_0}{s_X} \sqrt{\frac{N_1 N_0}{N(N-1)}}$$
where $\bar{X}_1, \bar{X}_0$ are the group means and $s_X$ is the sample standard deviation.
- **ANOVA $F$-statistic:** Determines whether the mean of a continuous feature (e.g., `danceability`) varies significantly across multiple categories (e.g., across 114 genres).

# Section 2: Principal Component Analysis (PCA)

## 1. Preprocessing Steps and Mathematical Formulation of PCA

### A. Necessary Preprocessing Steps Before Applying PCA
1. **Feature Selection:** PCA requires **continuous, quantitative numerical variables**. Categorical/identifier features (`track_id`, `artists`, `track_name`, `track_genre`) are excluded from PCA computation.
2. **Missing Value Imputation / Cleaning:** PCA cannot handle missing (`NaN`) values; the dataset must be completely clean.
3. **Standardization ($Z$-score scaling) — Strictly Mandatory:**
   $$\mathbf{Z} = \frac{\mathbf{X} - \mu}{\sigma}$$
   * **Why it is essential:** PCA finds directions (orthogonal axes) that maximize variance. If features are not standardized, a feature with large numerical scale (such as `duration_ms` with variance $\approx 10^{10}$) will completely dominate all principal components, while features with small variances (such as `danceability` with variance $\approx 0.03$) will be completely ignored, regardless of their actual information content.

### B. How Principal Components are Computed and Selected
1. **Covariance Matrix Calculation:**
   $$\mathbf{\Sigma} = \frac{1}{N-1} \mathbf{Z}^T \mathbf{Z}$$
2. **Eigendecomposition (or Singular Value Decomposition - SVD):**
   $$\mathbf{\Sigma} \mathbf{v}_i = \lambda_i \mathbf{v}_i$$
   - **Eigenvectors ($\mathbf{v}_i$):** Define the principal loading directions (linear combinations of original features).
   - **Eigenvalues ($\lambda_i$):** Measure the amount of variance captured along each principal component direction.
3. **Component Ordering and Selection:**
   - Eigenvalues are sorted in descending order: $\lambda_1 \ge \lambda_2 \ge \dots \ge \lambda_p$.
   - The top $k$ components are selected such that they satisfy a cumulative explained variance threshold (e.g., $\ge 80\%$ or $\ge 90\%$), or via the **Elbow Method** on a Scree Plot.
4. **Projection into Lower-Dimensional Subspace:**
   $$\mathbf{X}_{\text{pca}} = \mathbf{Z} \cdot \mathbf{W}_k$$
   where $\mathbf{W}_k = [\mathbf{v}_1, \mathbf{v}_2, \dots, \mathbf{v}_k]$ is the $p \times k$ projection matrix.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Select 10 continuous audio features
pca_features = [
    'duration_ms', 'danceability', 'energy', 'loudness', 'speechiness', 
    'acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo'
]

# 1. Standardize features (Mean = 0, Std = 1)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_clean[pca_features])

# 2. Compute full PCA
pca_full = PCA()
X_pca_full = pca_full.fit_transform(X_scaled)

# 3. Compute Explained Variance & Cumulative Variance
explained_var = pca_full.explained_variance_
explained_var_ratio = pca_full.explained_variance_ratio_ * 100
cumulative_var_ratio = np.cumsum(explained_var_ratio)

pca_summary_df = pd.DataFrame({
    'Eigenvalue (Variance)': np.round(explained_var, 3),
    'Explained Variance (%)': np.round(explained_var_ratio, 2),
    'Cumulative Variance (%)': np.round(cumulative_var_ratio, 2)
}, index=[f'PC{i+1}' for i in range(len(pca_features))])

print('=== PCA Explained Variance Table ===')
display(pca_summary_df)
```

### B. In-Depth Discussion: Advantages and Limitations of PCA

#### 1. Advantages of PCA
* **Complete Elimination of Multicollinearity:** Original features like `energy` and `loudness` had strong linear correlation ($r = +0.76$). PCA components are **strictly orthogonal (uncorrelated)** by mathematical design ($r_{\text{PC}_i, \text{PC}_j} = 0, \forall i \ne j$), resolving multicollinearity for linear regressors.
* **Dimensionality & Storage Reduction:** Reduces 10 features down to 6 principal components while retaining **83.5%** of the underlying variance, cutting down model training time and memory footprint.
* **Effective 2D/3D Visualization:** Allows high-dimensional musical profiles to be visually inspected in human-interpretable 2D/3D scatter plots.
* **Noise Filtering:** Discarding the lowest components (PC8–PC10, explaining < 10% variance) strips away recording sensor noise and random variance, often boosting the generalization capability of downstream classifiers.

#### 2. Limitations of PCA
* **Loss of Physical Interpretability:** In the original space, `tempo = 120 BPM` or `loudness = -5 dB` have clear, intuitive meanings. In the PCA space, `PC1 = 2.45` is an abstract linear combination of all 10 features, making it hard to explain predictions to music artists or business stakeholders.
* **Assumption of Linearity:** PCA assumes the underlying relationships between variables are linear. Non-linear manifolds (e.g., circular harmony loops or complex non-linear acoustic textures) are better captured by non-linear techniques like **t-SNE** or **UMAP**.
* **Information Loss:** Discarding components inevitably causes some information loss (e.g., subtle cues like `liveness` or niche vocal styles might be lost when keeping only top components).
* **High Sensitivity to Outliers and Scaling:** Because PCA maximizes variance, extreme uncleaned outliers or improperly scaled features distort the principal axes completely.
